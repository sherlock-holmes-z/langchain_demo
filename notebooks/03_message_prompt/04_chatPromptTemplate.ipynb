```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

api_key = os.getenv("DEEPSEEK_API_KEY")
base_url = os.getenv("DEEPSEEK_API_BASE")

chat_model = init_chat_model(api_key=api_key, base_url=base_url, model="deepseek-flash", model_provider="deepseek")
```

### ChatPromptTemplate
* 三种角色
    * system
    * ai /assistant
    * user / human

* MessagesPlaceholder("history") 插入历史消息

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([("system", "你是一个Java开发助手"), ("human", "{question}")])

result = prompt.invoke({"question": "1+1=?"})
print(result)
```

```python
from langchain_core.messages import AIMessage, HumanMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# 可以直接实例化，不调用from_messages方法；推荐
prompt = ChatPromptTemplate(
    [("system", "你是一个{language}开发助手"), MessagesPlaceholder("history"), ("human", "{question}")]
)

history = [
    HumanMessage(content="我叫张三"),
    AIMessage(content="你好，张三"),
    HumanMessage(content="我是一名 Java 开发"),
    AIMessage(content="好的，我记住了"),
]

# 返回消息体ChatPromptValue，推荐
chat_prompt = prompt.invoke(
    {"language": "python", "question": "我问的第一个问题是什么,你是什么开发助手", "history": history}
)
print(chat_prompt)
print(type(chat_prompt))
print(chat_prompt.to_messages())  # 转成消息list，同prompt.format_messages
print(chat_prompt.to_string())  # 转消息str,同prompt.format
print("===========================================")

# format 返回消息字符串
format_result = prompt.format(language="python", question="你是谁", history=history)
print(format_result)
print(type(format_result))
print("===========================================")

# format_messages 返回消息列表
from_messages_result = prompt.format_messages(language="python", question="你是谁", history=history)
for msg in from_messages_result:
    print(type(msg).__name__, ":", msg.content)
print(type(from_messages_result))
```

```python
response = chat_model.invoke(chat_prompt)
print(response.content)
print(response.content_blocks)
```
