```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage

load_dotenv(override=True)

api_key = os.getenv("DEEPSEEK_API_KEY")
base_url = os.getenv("DEEPSEEK_API_BASE")

chat_model = init_chat_model(api_key=api_key, base_url=base_url,model='deepseek:deepseek-flash')
```

### 多模态数据--使用字典列表
* 简单文本消息，直接使用content
* 多模态消息，使用content_blocks

```python
import base64
import mimetypes
from pathlib import Path

image_path = Path("../data/vuex.png")
mime_type, _ = mimetypes.guess_type(image_path)
mime_type = mime_type or "image/png"

image_base64 = base64.b64encode(image_path.read_bytes()).decode("ascii")

messages = [
    HumanMessage(
        content_blocks=[
            {"type": "text", "text": "一句话描述图片内容"},
            {"type": "image", "base64": image_base64, "mime_type": mime_type},
        ]
    )
]
# DeepSeek 官方文档明确说明deepseek-flash Thinking Mode 默认开启、默认强度为 high,
response = chat_model.invoke(messages)

print(response)
print("=========================")
print(response.content)
print("============思考过程统一封装在reasoning中=============")
print(response.content_blocks)
```
