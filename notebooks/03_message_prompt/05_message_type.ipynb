### ChatPromptTemplate的6重实例化方式
* str列表	                ⭐⭐	            简单场景	快速 Demo
* tuple列表	                ⭐⭐⭐⭐⭐	    最常用	普通业务开发
* dict列表	                ⭐⭐	            较少	动态配置 Prompt
* Message列表	            ⭐⭐⭐	            常用	历史消息、Agent
* BaseMessagePromptTemplate列表	⭐⭐⭐⭐⭐	大型项目常用	可复用 Prompt
* BaseChatPromptTemplate列表	⭐⭐⭐⭐	        高级扩展	自定义 Prompt

### 1.字符串列表
* 消息类型只有human,无法指定消息角色

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate(["你好，我是{name}"])
prompt = chat_prompt_template.invoke({'name':'zhangsan'})
print(prompt)
```

### 2.tuple列表
* 定位是Prompt模板
* 常用于system,human模版

```python
chat_prompt_template = ChatPromptTemplate([
    ("system", "你是一个{language}开发助手"),
    ("human", "{question}")
])
prompt = chat_prompt_template.invoke({'language': 'java', 'question': '你是谁'})
print(prompt)
```

### 3.字典列表
* 定位是消息数据，写法稍微复杂
* 常用于历史聊天内容等已确定的消息，或者数据库需要存储prompt,使用json可以直接存取

```python
chat_prompt_template = ChatPromptTemplate([
    {"role": "system","content": "你是{name}"},
    {'role': 'human','content': 'who are you'}
])
prompt = chat_prompt_template.invoke({'name': 'robot'})
print(prompt)
```

### 4.Message消息对象列表
* 消息对象中不可以声明变量
* Agent中大量使用

```python
chat_prompt_template = ChatPromptTemplate([
    SystemMessage(content='你是{name}'),  # name不会被替换
    HumanMessage(content='who are you'),
    AIMessage(content='i am robot')
])
prompt = chat_prompt_template.invoke({'name': 'robot'})
print(prompt)
```

### 5. BaseMessagePromptTemplate消息模版列表
* 大型项目使用，消息模板可以复用
* 消息中可以声明变量

```python
from langchain_core.prompts import SystemMessagePromptTemplate,HumanMessagePromptTemplate, AIMessagePromptTemplate

system_message_template =  SystemMessagePromptTemplate.from_template("你是{name}")
human_message_template = HumanMessagePromptTemplate.from_template('who are you')
ai_message_template = AIMessagePromptTemplate.from_template('i am robot')

chat_prompt_template = ChatPromptTemplate([
    system_message_template,
    human_message_template,
    ai_message_template
])
prompt = chat_prompt_template.invoke({'name': 'robot'})
print(prompt)
```

### 6.BaseChatPromptTemplate列表
* 嵌套ChatPromptTemplate
* 可以声明变量

```python
system_chat_prompt = ChatPromptTemplate([
    ('system','你是{name}')
])
human_chat_prompt = ChatPromptTemplate([
    ('human','你是谁')
])

chat_prompt_template = ChatPromptTemplate([
    system_chat_prompt,
    human_chat_prompt
])

prompt = chat_prompt_template.invoke({'name': 'robot'})
print(prompt)
```
