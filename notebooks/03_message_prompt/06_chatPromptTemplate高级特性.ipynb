#### 部分变量预填充partial()

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template =  ChatPromptTemplate([
    ('system','你是{name}'),
    ('user','{question}'),
])
# 预填充partial
final_template = chat_prompt_template.partial(name='robot')

prompt1 = final_template.invoke({'question':'who are you'})
prompt2 = final_template.invoke({'question':'how are you'})
print(prompt1)
print(prompt2)
```
